### Comment Writing Engine

#### an application to help teachers write better comments in less time



```python
# Import modules first!
from pathlib import Path
import csv
import os
```

Link to input Spreadsheet \(by Zach\): 

https://docs.google.com/spreadsheets/d/1RMVurab3czDBeOQPDB8XnjzNqVw68PkBjQko9iZvzpY/edit?usp=sharing 

The document name must remain 'Comments'. Each individual sheet should have the teacher's name that they wish to appear in the comments. When input is complete, the indivdual sheet must be exported as a CSV file and imported into the 'Teacher\_Comments' directory here. 

Used chat GPT to write sample comments.



```python
# functions


def get_teachers() -> list[list[str]]:
    """
    No inputs
    Requires in environment: a directory named "Teacher_Comments" with csv files
    for each teacher's comments from the spreadsheet.
    Returns all the comments left by every teacher for every student
    in the format [[student_name, *class_info], [[student_name, *class_info]...].
    
    Note: every student will have a seprate entry for every class the have.
    Empty cells won't be cleaned here!
    By: Jane
    """
    all_comments = []
    directory_path = Path("Teacher_Comments")

    # Iterate over all items in the directory
    for file_path in directory_path.iterdir():

        with open(file_path) as f:
            path = os.path.basename(
                file_path)  # e.g. "Teacher_Comments/Comments - PK.csv"
            teacher_name = path[11:-4]  # e.g. "PK"

            file = csv.reader(f, delimiter=',')
            header = next(file)  # Used to check index

            for line in file:
                all_comments.append(
                    line[:10] + [teacher_name])  # cleans out the helper cells

    return all_comments


def student_courses(
        all_comments: list[list[str]]) -> dict[str:list[list[str]]]:
    """
    Takes in list of lists that have all the comments in get_teachers() format.
    Returns dictionry with the keys being student name
    and values as list of lists of comments from all of thier classes.
    By: Zach & Jane
    """
    student_comments = {}

    for student in all_comments:
        name = student[0]
        if name != "":  # cleans the empty cells
            student_comments[name] = student_comments.get(name,
                                                          []) + [student[1:]]

    return student_comments
```

```python
# Main writing fun


def main():
    """
    Requires:
        functions: student_courses() and get_teachers()
        file: a folder with each teacher's comments
    Writes all students text files with each class comment entries.
    Each text file is in the directory 'Student_Comments'.
    By: Zach & Jane
    """
    student_comments = student_courses(get_teachers())

    for name, classes in student_comments.items():
        student_name = name

        file_path = f'Student_Comments/{student_name}.txt'
        if os.path.exists(file_path):
            os.remove(file_path)

        for class_details in classes:
            class_name = class_details[0]
            teacher = class_details[-1]

            strength_LO = class_details[4]
            strength_details = class_details[5]

            growth_LO = class_details[6]
            growth_details = class_details[7]

            summary = class_details[8]

            with open(f'Student_Comments/{student_name}.txt', 'a') as comment:
                comment.write(f'{class_name} - {teacher}\n\n')
                comment.write(f'Learning Outcome: {strength_LO}\n')
                comment.write(f'Stregths:\n\t{strength_details}\n\n')
                comment.write(f'Learning Outcome: {growth_LO}\n')
                comment.write(f'Areas for Growth:\n\t{growth_details}\n\n')
                comment.write(f'{summary}\n\n\n\n')


if __name__ == '__main__':
    main()
```
